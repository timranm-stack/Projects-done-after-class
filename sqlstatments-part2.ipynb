```python
import sqlite3
import pandas as pd
```

```python
# 1. Connect to an in-memory database (or a file like 'wildlife_park.db')
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
```

```python
# 2. Create the database table
cursor.execute('''
CREATE TABLE IF NOT EXISTS animals (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    species TEXT NOT NULL,
    habitat TEXT NOT NULL,
    age INTEGER,
    weight REAL
)
''')

# Insert sample wildlife records
sample_animals = [
    ('Leo', 'Lion', 'Savannah', 5, 190.5),
    ('Bella', 'Lion', 'Savannah', 3, 130.2),
    ('Pingu', 'Penguin', 'Polar', 2, 12.0),
    ('Chirpy', 'Penguin', 'Polar', 4, 15.5),
    ('Horton', 'Elephant', 'Savannah', 12, 4500.0),
    ('George', 'Monkey', 'Rainforest', 6, 25.0),
    ('Coco', 'Monkey', 'Rainforest', 1, 8.5),
    ('Aquila', 'Eagle', 'Mountains', 7, 6.2)
]

cursor.executemany('''
INSERT INTO animals (name, species, habitat, age, weight)
VALUES (?, ?, ?, ?, ?)
''', sample_animals)
conn.commit()

print("✅ Wildlife database created and populated successfully!\n")


# --- RUNNING THE ANALYZER ASSIGNMENT GOALS ---

print("=" * 60)
print("GOAL 1: Use DISTINCT to find unique values (Unique Habitats)")
print("=" * 60)
query_distinct = "SELECT DISTINCT habitat FROM animals;"
df_distinct = pd.read_sql_query(query_distinct, conn)
print(df_distinct, "\n")


print("=" * 60)
print("GOAL 2: Use ORDER BY to sort data (Heaviest to Lightest Animals)")
print("=" * 60)
query_order = "SELECT name, species, weight FROM animals ORDER BY weight DESC;"
df_order = pd.read_sql_query(query_order, conn)
print(df_order, "\n")


print("=" * 60)
print("GOAL 3: Use COUNT() and SUM() to total data")
print("=" * 60)
query_count_sum = "SELECT COUNT(*) as total_animals, SUM(weight) as total_weight_kg FROM animals;"
df_count_sum = pd.read_sql_query(query_count_sum, conn)
print(df_count_sum, "\n")


print("=" * 60)
print("GOAL 4: Use AVG() to calculate averages (Average Age of Animals)")
print("=" * 60)
query_avg = "SELECT AVG(age) as average_age_years FROM animals;"
df_avg = pd.read_sql_query(query_avg, conn)
print(df_avg, "\n")


print("=" * 60)
print("GOAL 5: Use GROUP BY to summarize records by habitat")
print("=" * 60)
query_group = """
SELECT habitat, 
       COUNT(*) as animal_count, 
       AVG(weight) as avg_weight_kg
FROM animals 
GROUP BY habitat;
"""
df_group = pd.read_sql_query(query_group, conn)
print(df_group, "\n")

# Close the database connection
conn.close()
```

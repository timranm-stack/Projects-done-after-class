```python
import sqlite3
import pandas as pd
```

```python
# 1. Create and connect to the SQLite database
conn = sqlite3.connect('library_finder.db')

# Create a sample table for books
query_create_table = """
CREATE TABLE IF NOT EXISTS books (
    book_id INTEGER PRIMARY KEY,
    title TEXT,
    author TEXT,
    genre TEXT,
    published_year INTEGER,
    price REAL
);
"""
conn.execute(query_create_table)

# Insert sample data into the table
sample_books = [
    (1, 'The Great Gatsby', 'F. Scott Fitzgerald', 'Fiction', 1925, 10.99),
    (2, 'To Kill a Mockingbird', 'Harper Lee', 'Fiction', 1960, 8.99),
    (3, 'A Brief History of Time', 'Stephen Hawking', 'Science', 1988, 14.99),
    (4, 'The Hobbit', 'J.R.R. Tolkien', 'Fantasy', 1937, 12.50),
    (5, '1984', 'George Orwell', 'Dystopian', 1949, 9.99),
    (6, 'Brave New World', 'Aldous Huxley', 'Dystopian', 1932, 11.25)
]

cursor = conn.cursor()
cursor.executemany("INSERT OR IGNORE INTO books VALUES (?, ?, ?, ?, ?, ?);", sample_books)
conn.commit()

print("--- Database Created and Data Inserted Successfully ---")
```

```python
# 2. Use SELECT to view all columns or specific columns
print("\n--- Goal: SELECT all columns ---")
df_all = pd.read_sql_query("SELECT * FROM books;", conn)
print(df_all)

print("\n--- Goal: SELECT specific columns (Title and Author) ---")
df_specific = pd.read_sql_query("SELECT title, author FROM books;", conn)
print(df_specific)
```

```python
# 3. Use WHERE to filter records matching a condition
print("\n--- Goal: WHERE condition (Books published after 1950) ---")
df_filtered = pd.read_sql_query("SELECT * FROM books WHERE published_year > 1950;", conn)
print(df_filtered)
```

```python
# 4. Use LIKE to search for patterns in book titles
print("\n--- Goal: LIKE pattern search (Titles starting with 'The') ---")
df_pattern = pd.read_sql_query("SELECT * FROM books WHERE title LIKE 'The%';", conn)
print(df_pattern)
```

```python
# 5. Use MIN() and MAX() to find smallest and largest values
print("\n--- Goal: MIN() and MAX() functions (Price extremes) ---")
df_min_max = pd.read_sql_query("""
    SELECT MIN(price) AS lowest_price, MAX(price) AS highest_price 
    FROM books;
""", conn)
print(df_min_max)

# Close the database connection when finished
conn.close()
```
